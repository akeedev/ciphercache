# Secret Envelope (070) Demo

Demonstrate Secret and SecretEnvelope redaction and usage helpers.


```python
%load_ext autoreload
%autoreload 2

from ciphercache.secret import Secret, SecretEnvelope
```

```python
secret = Secret("s3cret")
print(secret)
print(secret.reveal())

envelope = SecretEnvelope.from_payload({"user": "alice", "password": "s3cret"})
print(envelope)
result = envelope.use(["user", "password"], lambda user, password: f"{user}:{password}")
print(result)
```
