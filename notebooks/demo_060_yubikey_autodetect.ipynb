# YubiKey Autodetect (spec 060)

Demonstrates the optional YubiKey autodetection module. This notebook uses
real `ykman` output (if installed) to detect attached YubiKeys. If `ykman`
is not installed, the availability check will return `False` and detection
will raise an error.

```python
%load_ext autoreload
%autoreload 2

from ciphercache.yubikey import detect_yubikey, is_ykman_available
```

```python
# Check whether ykman is available in PATH
available = is_ykman_available()
print(f"ykman available: {available}")
```

```python
# Attempt autodetection (requires ykman and a connected YubiKey)
if available:
    try:
        serial = detect_yubikey()
        print(f"Detected YubiKey serial: {serial}")
    except RuntimeError as exc:
        print(f"Detection failed: {exc}")
else:
    print("Skipping detection: ykman is not installed.")
    print("Install via: brew install ykman")
```

```python
# Using autodetect with the daemon runner:
# Start the daemon with --yubikey auto to enable autodetection.
#
#   uv run python scripts/run_daemon.py \
#     --db-path testdata/demopasswords.kdbx \
#     --yubikey auto
#
# The daemon will call detect_yubikey() during startup unlock and use
# the detected serial with slot 1 (i.e. --yubikey 1:<serial>).
print("See daemon runner --yubikey auto for integrated usage.")
```
